```python
%load_ext autoreload
%autoreload 2
%matplotlib widget
import pandas as pd
import os
from core.helpers import (get_config_file, get_data_out, get_data_computed_in, get_experiment_id)
PROJECT_FILE = 'project_ID8_5.yaml'
CASE = 'JP1'

ROOT = os.path.dirname(os.path.dirname(os.getcwd()))
DATA_IN = os.path.join(ROOT, 'data', CASE, 'Input_Data')
CONFIG = get_config_file(DATA_IN, PROJECT_FILE)

DATA_COMPUTED_IN = get_data_computed_in(DATA_IN, PROJECT_FILE)
DATA_OUT         = get_data_out(DATA_IN, PROJECT_FILE)

print(f"Experiment ID: '{get_experiment_id(DATA_IN, PROJECT_FILE)}'")
```

```python
import itertools 

hp_dict = {
    'N_nodes': [64, 128, 256],
    'N_layers': [2, 4, 8],
    'batch_size': [256, 1024],
    'lr': [1e-4, 1e-5, 1e-6],
    'weight_decay': [1e-4, 1e-5, 1e-6]
}

# Generate all combinations
keys = hp_dict.keys()
values = hp_dict.values()
combinations = list(itertools.product(*values))

# Build DataFrame and make sure, the N_nodes are saved as integers
df_hp = pd.DataFrame(combinations, columns=keys)
# df_hp['N_nodes'] = df_hp['N_nodes'].astype(int)

n_epochs = 300
estimated_time_per_epoch = 0.1

print(f"estimated total time : {n_epochs * len(df_hp) * estimated_time_per_epoch/3600:.2f} h.")

df_hp.to_csv(os.path.join(DATA_COMPUTED_IN, 'hp_tuning.csv'))
```
